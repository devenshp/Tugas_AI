# TOPIK 1: PEMETAAN VEKTOR KECEPATAN ANGIN REGIONAL

### A. Latar Belakang Teori & Konsep Fisis
Komponen arah dan kecepatan angin pada sistem koordinat spasial dihitung menggunakan komponen $U$ (zonal/timur-barat) dan $V$ (meridional/utara-selatan):
- **Kecepatan Angin Total ($S$):**
  $$S = \sqrt{u^2 + v^2}$$
- **Arah Vektor Angin:**
  $$\theta = \arctan2(v, u)$$
Visualisasi spasial berbasis GIS memungkinkan analisis pola sirkulasi angin regional dan magnitudo kecepatan angin pada grid spasial tertentu.

### B. Prompt AI (Struktur R–T–C–I–K–O–Q–V)
```markdown
# R — Role
Anda berperan sebagai ahli GIS dan Meteorologi/Geofisika yang membimbing mahasiswa S1 Geofisika dalam visualisasi data vektor spasial.

# T — Task
Buat implementasi Python (menggunakan matplotlib) untuk memplot vektor angin regional dari data komponen u dan v, menghitung magnitudo kecepatan total, serta menampilkan visualisasi peta berkolom warna (colormap) dan vektor panah yang informatif.

# C — Context
Pengolahan data angin regional dilakukan untuk visualisasi peta kecepatan angin pada tanggal tertentu sebagai bagian dari proyek peta GIS geofisika.

# I — Input
Grid koordinat bujur (lon), lintang (lat), serta komponen kecepatan angin u dan v.

# K — Constraints
- Gunakan formula fisika yang tepat untuk magnitudo vektor: speed = sqrt(u^2 + v^2).
- Tampilkan panah arah vektor (quiver plot) dengan skala yang proporsional di atas peta berkolom warna.
- Berikan label sumbu koordinat geografis (Bujur dan Lintang) secara jelas.
- Sertakan colorbar untuk indikasi kecepatan angin dalam m/s.

# O — Output
1. Penjelasan ringkas konversi komponen u dan v menjadi magnitudo.
2. Hasil perhitungan statistik magnitudo.
3. Grafik peta spasial vektor angin lengkap dengan colormap dan quiver arrows.
4. Interpretasi fisik pola sirkulasi angin.

# =============================================================================
# KODE PYTHON: PEMETAAN VEKTOR KECEPATAN ANGIN REGIONAL (GIS & GEOFISIKA)
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt

# 1. Inisialisasi Grid Koordinat dan Komponen Vektor Angin (u, v)
# Bujur (lon) dari 118°E hingga 122°E, Lintang (lat) dari -6°S hingga -2°S
lon = np.linspace(118, 122, 5)
lat = np.linspace(-6, -2, 5)
LON, LAT = np.meshgrid(lon, lat)

# Matriks komponen kecepatan zonal (u) dan meridional (v) dalam m/s
u = np.array([[2, 3, 4, 3, 2], 
              [3, 4, 5, 4, 3], 
              [4, 5, 6, 5, 4], 
              [3, 4, 5, 4, 3], 
              [2, 3, 4, 3, 2]])

v = np.array([[1, 2, 2, 1, 0], 
              [2, 3, 3, 2, 1], 
              [3, 4, 4, 3, 2], 
              [2, 3, 3, 2, 1], 
              [1, 2, 2, 1, 0]])

# 2. Perhitungan Magnitudo Kecepatan Angin Total (S = sqrt(u^2 + v^2))
speed = np.sqrt(u**2 + v**2)

# Menampilkan hasil statistik komputasi ke konsol
print("="*55)
print("  HASIL ANALISIS VEKTOR KECEPATAN ANGIN REGIONAL")
print("="*55)
print(f"Kecepatan Angin Minimum : {np.min(speed):6.2f} m/s")
print(f"Kecepatan Angin Maksimum : {np.max(speed):6.2f} m/s")
print(f"Rata-rata Kecepatan     : {np.mean(speed):6.2f} m/s")
print("="*55)

# 3. Visualisasi Spasial Plotting (Contourf + Quiver Plot)
plt.figure(figsize=(9, 6), dpi=100)

# Membuat kontur warna (colormap) untuk magnitudo kecepatan angin
contour = plt.contourf(LON, LAT, speed, cmap='YlGnBu', levels=15, alpha=0.85)
cbar = plt.colorbar(contour)
cbar.set_label('Kecepatan Angin Total (m/s)', fontsize=11, fontweight='bold')

# Menambahkan panah vektor arah angin (quiver plot)
quiver_plot = plt.quiver(LON, LAT, u, v, color='black', scale=30, width=0.005, headwidth=4, label='Vektor Angin (u, v)')

# Pengaturan Layout dan Label Peta GIS
plt.title('Peta Spasial Vektor Kecepatan Angin Regional', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Bujur (°E)', fontsize=11)
plt.ylabel('Lintang (°S)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6, color='gray')
plt.legend(loc='lower right', frameon=True, facecolor='white', edgecolor='none')
plt.tight_layout()

# Menampilkan Plot
plt.show()

## 4. PENJELASAN KELUARAN (*OUTPUT*) DAN INTERPRETASI FISIS

Setelah kode di atas dijalankan melalui lingkungan interaktif di VS Code, program menghasilkan dua jenis keluaran utama:

### A. Keluaran Konsol (*Console Output*)
* **Anomali Bouguer Min/Max:** Memberikan rentang nilai percepatan gravitasi terkoreksi di area survei (mGal). Rentang ini menunjukkan variasi kontras densitas batuan regional.
* **Nilai HGM Maksimum:** Menunjukkan intensitas gradien tertinggi. Nilai puncak ini menjadi indikator kuantitatif utama adanya kelurusan struktur geologi atau bidang sesar tersembunyi (*blind fault*).

### B. Keluaran Visualisasi Grafik (*Spatial Plots*)
1. **Subplot 1 (Peta Anomali Bouguer & Vektor Sesar):**
   * *Warna Kontur (Colormap 'terrain'):* Menampilkan variasi nilai anomali. Zona berwarna hijau/biru tua merepresentasikan daerah beranomali rendah (diduga cekungan sedimen), sedangkan zona cokelat/putih merepresentasikan anomali tinggi (intrusi batuan beku atau pengangkatan basement).
   * *Panah Merah (Quiver):* Menunjuk ke arah perubahan nilai kontras densitas maksimum. Garis panah yang terkonsentrasi secara linier menandakan jalur zona lemah atau patahan geologi.
2. **Subplot 2 (Peta HGM / Edge Detection):**
   * *Skala Warna Hangat ('hot_r'):* Menyoroti area dengan gradien tajam secara spasial. Area berwarna gelap atau merah menyala menunjukkan lokasi ketidakontinuitasan struktur bawah permukaan secara akurat berdasarkan prinsip pemrosesan citra GIS.

## 6. TOPIK 5: ANALISIS KERAPATAN KELURUSAN (LINEAMENT DENSITY) UNTUK IDENTIFIKASI STRUKTUR

### A. Latar Belakang Teori & Konsep Fisis
Analisis kerapatan kelurusan (*lineament density*) merupakan salah satu metode ekstraksi fitur linier dari data citra penginderaan jauh atau peta DEM (*Digital Elevation Model*). Kelurusan permukaan sering kali mencerminkan ekspresi morfologi dari struktur geologi bawah permukaan seperti sesar, kekar, atau zona rekahan. 
- **Definisi Kerapatan Kelurusan ($LD$):**
  $$LD = \frac{\sum L_i}{A}$$
  Dimana $\sum L_i$ adalah total panjang segmen kelurusan dalam suatu sel grid tertentu, dan $A$ adalah luas area grid tersebut.
- *Konsep Fisis:* Nilai kerapatan kelurusan yang tinggi pada suatu wilayah menunjukkan intensitas deformasi tektonik yang kuat, yang sangat penting dalam pemetaan bahaya geologi dan eksplorasi hidrokarbon atau panas bumi.

---

### B. Alur Metodologi Analisis Kerapatan Kelurusan
1. **Ekstraksi Fitur Linier:** Mengidentifikasi kelurusan morfologi dari peta DEM atau citra satelit secara visual maupun otomatis menggunakan algoritma deteksi tepi (*edge detection*).
2. **Pembuatan Grid Analisis (*Grid Mesh*):** Membagi wilayah studi ke dalam kotak-kotak grid dengan ukuran sel tertentu (misalnya $1 \times 1\text{ km}$).
3. **Kalkulasi Panjang Kumulatif:** Menghitung total panjang segmen kelurusan yang berada di dalam setiap sel grid untuk mendapatkan nilai densitas spasial.

---

### C. Penjelasan Keluaran (*Output*) dan Interpretasi Fisis
* **Peta Sebaran Segmen Kelurusan:** Menampilkan orientasi dan sebaran spasial garis-garis kelurusan untuk mendeterminasi tren struktur regional.
* **Peta Kerapatan Kelurusan (Density Contour):** Menunjukkan konsentrasi rekahan. Area dengan kerapatan tinggi menandakan zona lemah yang rentan terhadap pergerakan fluida atau aktivitas kegempaan.

---

## 7. TOPIK 6: KLASIFIKASI PENUTUP LAHAN BERBASIS DRONE UNTUK SURVEI GEOFISIKA LINGKUNGAN

### A. Latar Belakang Teori & Konsep Fisis
Pemetaan survei drone (*Unmanned Aerial Vehicle* / UAV) menghasilkan ortofoto beresolusi sangat tinggi. Klasifikasi penutupan lahan (*land cover classification*) diperlukan untuk mengidentifikasi geomorfologi permukaan serta mengoreksi faktor gangguan lingkungan yang mempengaruhi data ukur geofisika dekat permukaan (*near-surface geophysics*).
- **Indeks Spektral (seperti NDVI):** 
  $$NDVI = \frac{NIR - Red}{NIR + Red}$$
  Digunakan untuk membedakan kerapatan vegetasi, tanah terbuka (*bare soil*), dan infrastruktur bangunan secara spasial dalam sistem koordinat GIS.

---

### B. Alur Metodologi Klasifikasi Spasial
1. **Akuisisi Data Ortofoto:** Pengumpulan citra udara resolusi tinggi hasil survei drone lintas area target.
2. **Ekstraksi Fitur & Band Kalkulasi:** Memproses kanal warna RGB atau multispektral untuk menghitung indeks vegetasi dan karakteristik tekstur permukaan.
3. **Klasifikasi Tematik:** Mengelompokkan piksel citra ke dalam kelas penutupan lahan (vegetasi, bangunan, badan air, tanah terbuka) guna mendukung analisis tata guna lahan terpadu.

---

### C. Penjelasan Keluaran (*Output*) dan Interpretasi Fisis
* **Peta Klasifikasi Tematik:** Menyajikan zonasi penutupan lahan secara spasial di area survei.
* **Korelasi Lingkungan Geofisika:** Membantu dalam mengidentifikasi sumber *noise* permukaan (seperti aktivitas antropogenik atau vegetasi lebat) saat merancang layout titik ukur metode geofisika.
